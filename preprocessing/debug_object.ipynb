```python
import torch
import os

# Select a specific .pt file
pt_file = "/media/popsatorn/external/depressionFinalProject/DAIC-WOZ/DAIC_embeddings/validate/492.pt"  # Replace with your .pt file name

# Load the file
data = torch.load(pt_file, map_location='cpu')

# Print the type of the loaded data
print(f"Type of loaded data: {type(data)}")

# If it's a dictionary, display keys and their corresponding data info
if isinstance(data, dict):
    print("\nDictionary contents:")
    for key, value in data.items():
        print(f"\nKey: {key}")
        print(f"Type: {type(value)}")
        if isinstance(value, torch.Tensor):
            print(f"Shape: {value.shape}")
            print(f"Dtype: {value.dtype}")
            print(f"Device: {value.device}")
# If it's a tensor, display its properties
elif isinstance(data, torch.Tensor):
    print("\nTensor properties:")
    print(f"Shape: {data.shape}")
    print(f"Dtype: {data.dtype}")
    print(f"Device: {data.device}")
```

```python
# Display clnf_cols contents
print("CLNF Columns:")
for key, value in data['clnf_cols'].items():
    print(f"\n{key}:")
    print(value)

print("\nCLNF Data:")
for key, value in data['clnf'].items():
    print(f"\n{key}:")
    if isinstance(value, torch.Tensor):
        print(f"Shape: {value.shape}")
        print(f"Type: {value.dtype}")
        print(f"First few values: {value[:5]}")
    else:
        print(f"Type: {type(value)}")
        print(f"First few values: {value[:5]}")
```
